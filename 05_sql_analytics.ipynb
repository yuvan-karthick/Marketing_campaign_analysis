```python
import sqlite3
import pandas as pd

# Connect to existing database
conn = sqlite3.connect("C:/Users/yuvan/OneDrive/Desktop/ML_projects/Marketing_campaign_analysis/data/marketing_analytics.db")
```

```python
pd.read_sql("SELECT COUNT(*) FROM customers", conn)

```

```python
pd.read_sql("""
SELECT 
    COUNT(*) AS total_customers,
    ROUND(AVG(Response) * 100, 2) AS overall_response_rate_pct
FROM customers
""", conn)
```

```python
pd.read_sql("""
SELECT 
    Customer_Segment,
    COUNT(*) AS customers,
    ROUND(AVG(Response) * 100, 2) AS response_rate_pct
FROM customers
GROUP BY Customer_Segment
ORDER BY response_rate_pct DESC
""", conn)
```

```python
pd.read_sql("""
SELECT
    ROUND(AVG(AcceptedCmp1) * 100, 2) AS Cmp1_Response,
    ROUND(AVG(AcceptedCmp2) * 100, 2) AS Cmp2_Response,
    ROUND(AVG(AcceptedCmp3) * 100, 2) AS Cmp3_Response,
    ROUND(AVG(AcceptedCmp4) * 100, 2) AS Cmp4_Response,
    ROUND(AVG(AcceptedCmp5) * 100, 2) AS Cmp5_Response
FROM customers
""", conn)
```

```python
pd.read_sql("""
SELECT 
    Customer_Segment,
    ROUND(AVG(Total_Spend), 2) AS avg_spend,
    ROUND(AVG(Income), 2) AS avg_income,
    ROUND(AVG(Response) * 100, 2) AS response_rate_pct
FROM customers
GROUP BY Customer_Segment
ORDER BY avg_spend DESC
""", conn)
```

```python
pd.read_sql("""
SELECT 
    Customer_Segment,
    ROUND(AVG(NumWebPurchases), 2) AS avg_web_purchases,
    ROUND(AVG(NumCatalogPurchases), 2) AS avg_catalog_purchases,
    ROUND(AVG(NumStorePurchases), 2) AS avg_store_purchases
FROM customers
GROUP BY Customer_Segment
ORDER BY avg_store_purchases DESC
""", conn)
```

```python
pd.read_sql("""
SELECT 
    Customer_Segment,
    COUNT(*) AS customers,
    ROUND(AVG(NumWebVisitsMonth), 2) AS avg_web_visits,
    ROUND(AVG(Response) * 100, 2) AS response_rate_pct,
    ROUND(AVG(Total_Spend), 2) AS avg_spend
FROM customers
GROUP BY Customer_Segment
HAVING avg_web_visits > 5 AND response_rate_pct < 10
ORDER BY avg_web_visits DESC
""", conn)
```
