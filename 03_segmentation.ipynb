```python
import pandas as pd

df = pd.read_csv("C:/Users/yuvan/OneDrive/Desktop/ML_projects/Marketing_campaign_analysis/data/marketing_data_cleaned.csv")
```

```python
high_spend_threshold = df['Total_Spend'].quantile(0.90)
```

```python
def assign_segment(row):
    if row['Total_Spend'] >= high_spend_threshold:
        return 'High Spender'
    elif row['Income'] > 75000:
        return 'High Income'
    elif row['Response'] == 1:
        return 'Campaign Responder'
    elif row['NumWebVisitsMonth'] > 5:
        return 'High Web Engagement'
    elif row['Children'] > 0:
        return 'Family Customer'
    elif row['Age'] < 30:
        return 'Young Customer'
    else:
        return 'Standard Customer'
```

```python
df['Customer_Segment'] = df.apply(assign_segment, axis=1)
```

```python
df['Customer_Segment'].value_counts()
```

```python
segment_summary = (
    df.groupby('Customer_Segment')
      .agg(
          customers=('ID', 'count'),
          avg_income=('Income', 'mean'),
          avg_spend=('Total_Spend', 'mean'),
          response_rate=('Response', 'mean'),
          avg_web_visits=('NumWebVisitsMonth', 'mean')
      )
      .sort_values('avg_spend', ascending=False)
)

segment_summary
```
